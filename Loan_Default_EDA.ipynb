```python
# Feature Engineering: Replicating DAX Categorical Bands

# Create Income Band
income_conditions = [
    (df['Income'] < 20000),
    (df['Income'] >= 20000) & (df['Income'] < 50000),
    (df['Income'] >= 50000) & (df['Income'] < 100000),
    (df['Income'] >= 100000)
]
income_choices = ['Low', 'Medium', 'High', 'Very High']
df['Income Band'] = np.select(income_conditions, income_choices, default='Unknown')
```

```python
# Create Credit Score Band
credit_conditions = [
    (df['CreditScore'] < 580),
    (df['CreditScore'] >= 580) & (df['CreditScore'] < 670),
    (df['CreditScore'] >= 670) & (df['CreditScore'] < 740),
    (df['CreditScore'] >= 740) & (df['CreditScore'] < 800),
    (df['CreditScore'] >= 800)
]
```

```python
# Ordering the choices logically for later plotting
credit_choices = ['Very Poor', 'Fair', 'Good', 'Very Good', 'Excellent']
df['Credit Score Band'] = np.select(credit_conditions, credit_choices, default='Unknown')
```

```python
# Convert to categorical with ordered levels for cleaner charts
df['Income Band'] = pd.Categorical(df['Income Band'], categories=['Low', 'Medium', 'High', 'Very High'], ordered=True)
df['Credit Score Band'] = pd.Categorical(df['Credit Score Band'], categories=['Very Poor', 'Fair', 'Good', 'Very Good', 'Excellent'], ordered=True)
```

```python
# ---------------------------------------------------------
# Testing Assumptions (EDA)
# ---------------------------------------------------------
sns.set_theme(style="whitegrid")

# Assumption 1: Low Income + Poor Credit = High Default
plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='Credit Score Band', y='Default', hue='Income Band', errorbar=None, palette='viridis')
plt.title('Default Rate by Credit Score and Income Band', fontsize=14)
plt.ylabel('Default Rate')
plt.xlabel('Credit Score Band')
plt.legend(title='Income Band')
plt.show()

# Approved (but borrowers with above fair, good, very good, and excellent credit score are also likely to default if their income is low)
```

```python
# Assumption 2: Impact of Co-Signer Absence
plt.figure(figsize=(8, 5))
co_signer_rates = df.groupby('HasCoSigner')['Default'].mean().reset_index()
sns.barplot(data=co_signer_rates, x='HasCoSigner', y='Default', palette='mako')
plt.title('Default Rate Variance: Presence vs. Absence of Co-Signer', fontsize=14)
plt.ylabel('Average Default Rate')
plt.xlabel('Has Co-Signer?')
for index, row in co_signer_rates.iterrows():
    plt.text(row.name, row.Default, f'{row.Default:.2%}', color='black', ha="center", va="bottom")
plt.show()

#Approved (chanches of defaults are slightly high in absance of co-signer)
```

```python
# Assumption 3: Unemployment Risk by Loan Purpose
plt.figure(figsize=(12, 6))
sns.barplot(data=df, x='LoanPurpose', y='Default', hue='EmploymentType', errorbar=None, palette='magma')
plt.title('Default Rate by Loan Purpose and Employment Type', fontsize=14)
plt.ylabel('Default Rate')
plt.xlabel('Loan Purpose')
plt.legend(title='Employment Type', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

#approved borrowers are likely to default if they are unemployed followed by part timers
```

```python
# Assumption 4: DTI Ratio Correlation with Defaults
# Creating DTI Bins for visualization
df['DTI_Bin'] = pd.qcut(df['DTIRatio'], q=5, labels=['Lowest 20%', 'Lower-Mid 20%', 'Middle 20%', 'Upper-Mid 20%', 'Highest 20%'])

plt.figure(figsize=(10, 6))
sns.barplot(data=df, x='DTI_Bin', y='Default', errorbar=None, palette='rocket')
plt.title('Default Rate Trends Across Debt-to-Income (DTI) Quintiles', fontsize=14)
plt.ylabel('Default Rate')
plt.xlabel('DTI Ratio Quintiles')
plt.show()

#approved High dti ration lead to higher defaults
```
